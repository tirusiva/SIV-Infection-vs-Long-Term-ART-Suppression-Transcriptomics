```python
# =========== Pathway × Gene heatmap by pathway z-score ===========

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
from io import StringIO
from matplotlib.ticker import FuncFormatter

# ---- 1) Paste your table here ----
raw = r"""
Ingenuity Canonical Pathways	z-score	Molecules
S100 Family Signaling Pathway	-2.828	AVPR2,BDKRB2,CACNA1B,CELSR1,CHRM3,CXCR1,DEFA1B,FZD10,MMP25,MMP9
EPH-Ephrin signaling	-2.000	CLTCL1,EPHA1,EPHA2,MMP9
Neutrophil degranulation	-2.646	CDA,CXCR1,JUP,MGAM,MME,MMP25,MMP9
NF1 RAS Signaling Pathway	-2.646	AVPR2,BDKRB2,CELSR1,CHRM3,CSF3R,CXCR1,FZD10
Breast Cancer Regulation by Stathmin1	-2.449	AVPR2,BDKRB2,CELSR1,CHRM3,CXCR1,FZD10,MMP9
CREB Signaling in Neurons	-2.449	AVPR2,BDKRB2,CACNA1B,CELSR1,CHRM3,CXCR1,FZD10
Molecular Mechanisms of Cancer	-2.646	AVPR2,BDKRB2,CELSR1,CHRM3,CXCR1,FZD10,MMP25,MMP9
Zn Homeostasis Signaling Pathway	-2.828	AVPR2,BDKRB2,CELSR1,CHRM3,CXCR1,FZD10,MMP25,MMP9,MS4A2
BBSome Signaling Pathway	-2.236	AVPR2,BDKRB2,CELSR1,CHRM3,CXCR1,FZD10
Class A/1 (Rhodopsin-like receptors)	-2.236	AVPR2,BDKRB2,CHRM3,CXCR1,PROK2
Cellular Effects of Sildenafil (Viagra)	2.449	AVPR2,BDKRB2,CACNA1B,CELSR1,CHRM3,CXCR1,FZD10
Osteoarthritis Pathway	-2.000	FZD10,ITLN1,MMP9,PTGS2
Colorectal Cancer Metastasis Signaling	-2.000	FZD10,MMP25,MMP9,PTGS2
Phagosome Formation	-2.236	AVPR2,BDKRB2,CELSR1,CHRM3,CXCR1,FZD10
G-Protein Coupled Receptor Signaling	-2.236	AVPR2,BDKRB2,CELSR1,CHRM3,CXCR1,FZD10
FAK Signaling	-2.449	AVPR2,BDKRB2,CELSR1,CHRM3,CXCR1,FZD10,MMP9
"""

# ---- 2) Read data ----
df = pd.read_csv(StringIO(raw), sep="\t")
df["z-score"] = pd.to_numeric(df["z-score"], errors="coerce")

# ---- 3) Parse genes and build matrix ----
def split_genes(s):
    if pd.isna(s):
        return []
    return [g.strip() for g in re.split(r"[;,]", str(s)) if g.strip()]

df["Gene_list"] = df["Molecules"].apply(split_genes)

# preserve gene order by first appearance
all_genes = []
for genes in df["Gene_list"]:
    for g in genes:
        if g not in all_genes:
            all_genes.append(g)

mat = pd.DataFrame(
    np.nan,
    index=df["Ingenuity Canonical Pathways"],
    columns=all_genes
)

for _, r in df.iterrows():
    for g in r["Gene_list"]:
        mat.loc[r["Ingenuity Canonical Pathways"], g] = r["z-score"]

# ---- 4) Color scaling ----
vals = mat.to_numpy(dtype=float)
finite = vals[np.isfinite(vals)]
vmax = max(2.0, np.nanmax(np.abs(finite))) if finite.size else 2.0

# ---- 5) Plot ----
fig_h = max(7, len(mat) * 0.45)
fig_w = max(14, len(all_genes) * 0.45)

plt.figure(figsize=(fig_w, fig_h))

ax = sns.heatmap(
    mat,
    cmap="bwr",
    center=0,
    vmin=-vmax,
    vmax=vmax,
    cbar_kws={"label": "Pathway z-score"},
    linecolor="lightgrey",
    linewidths=0.3
)

ax.figure.axes[-1].yaxis.set_major_formatter(
    FuncFormatter(lambda x, pos: f"{x:.1f}")
)

ax.set_title(
    "Canonical Pathways × Gene Heatmap",
    fontsize=16,
    fontweight="bold",
    pad=12
)

ax.set_xlabel("Genes", fontsize=13, fontweight="bold")
ax.set_ylabel("Ingenuity Canonical Pathways", fontsize=13, fontweight="bold")

ax.set_xticklabels(
    ax.get_xticklabels(),
    rotation=90,
    fontsize=10,
    fontweight="bold"
)

ax.set_yticklabels(
    ax.get_yticklabels(),
    rotation=0,
    fontsize=10,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    "canonical_pathways_gene_heatmap_zscore.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    "canonical_pathways_gene_heatmap_zscore.pdf",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
